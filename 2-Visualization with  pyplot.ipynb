```python
from matplotlib import pyplot as plt
import numpy as np
```

```python
x = [1, 2, 3, 4]
y = [1, 4, 9, 16]

plt.plot(x, y)
plt.axis((0, 6, 0, 20))
plt.xticks(x)
plt.yticks(y)
plt.show()
```

```python
t = np.arange(0., 5., 0.2)


plt.plot(t, t, color = "r", label = "t", linewidth = 1)
plt.plot(t, t**2, color = "g", label = "t**2", linewidth = 1.5)
plt.plot(t, t**3, color = "purple", label = "t**3", linewidth = 2)
plt.legend()
plt.title("an example", fontsize = 15)
plt.show()
```

```python
names = ['group_a', 'group_b', 'group_c']
values = [10, 50, 100]

plt.figure(figsize=(4,4))

bars = plt.bar(names, values)
bars[0].set_hatch("o")
bars[1].set_hatch("/")
bars[2].set_hatch("*")

plt.title("bar chart")

plt.show()
```

```python
labels = ["boys" , "girls"]
sizes = np.array([30 , 7])
colors = [ "green", "purple"]

plt.pie(sizes , labels=labels, colors=colors, autopct="%1.1f%%", startangle=90, wedgeprops={"width" : 0.55, "edgecolor" : "white"})
plt.title("Khu 1403 CS")
plt.axis("equal")
plt.show()
```

```python
scores = np.random.normal(9 ,3 ,40)

plt.figure(figsize=(8,5))
plt.hist(scores, bins=8, color="skyblue", label="AP scores")
plt.legend()
plt.xlabel("score")
plt.ylabel("no of student")
plt.grid(axis="y")

plt.show()
```
